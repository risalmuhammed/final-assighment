```python
# SECTION 4: DATA WRANGLING

# Create an engagement score
social_df['engagement_score'] = (
    social_df['likes'] +
    social_df['comments'] +
    social_df['shares']
)

# Display the new engagement score
print("Engagement Score:")
print(social_df[['likes', 'comments', 'shares', 'engagement_score']].head())
```

```python
# Fill missing numerical values with the median
numeric_cols = ['age', 'likes', 'comments', 'shares']

for col in numeric_cols:
    social_df[col] = social_df[col].fillna(social_df[col].median())

# Fill missing categorical values with the mode
categorical_cols = ['gender', 'sentiment']

for col in categorical_cols:
    social_df[col] = social_df[col].fillna(social_df[col].mode()[0])

# Check missing values after cleaning
print("Missing Values After Cleaning:")
print(social_df.isnull().sum())

print("\nTotal Missing Values:", social_df.isnull().sum().sum())
```

# Social Media Engagement Analysis – Final Summary

## Introduction
This project analyzes social media engagement data containing 5,000 records. Python libraries such as Pandas, NumPy, Matplotlib and Seaborn were used for data cleaning, analysis and visualization.

## Key Findings
- Video posts recorded the highest average engagement score.
- France recorded the highest average likes.
- Mobile users showed the highest average engagement rate.
- Negative sentiment recorded the highest average engagement score.
- Engagement patterns were compared across post types, countries, devices and sentiments.

## Conclusion
The analysis shows that social media engagement varies across post types, countries, devices and sentiment categories. Video content performed best based on the average engagement score, while mobile users showed the highest engagement rate.
